# LangChain调用本地大模型

第一步：下载安装ollama

第二步：在ollama下部署大模型

第三步：在LangChain中调用ollama下的大模型

```python
from langchain_ollama import ChatOllama

# 调用本地大模型，省略api_key base_url
llm = ChatOllama(
    model = "qwen3.5:2b"
)
response = llm.invoke("今天星期几？")
print(response)
```

```python
from langchain_ollama import ChatOllama

# 调用本地大模型，省略api_key base_url
llm = ChatOllama(
    model = "deepseek-r1:8b"
)
response = llm.invoke("自我介绍一下")
print(response.content)
```
